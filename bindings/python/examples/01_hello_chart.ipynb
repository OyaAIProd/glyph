# Hello, Glyph

This notebook is the canonical 60-second tour: build a pandas DataFrame, hand
it to Glyph, and render an inline SVG chart. No file I/O, no boilerplate, no
network calls.

If you can run this top-to-bottom you have a working Glyph install.

```python
import pandas as pd
import glyph

df = pd.DataFrame({
    "hour": list(range(24)),
    "rides": [40, 35, 30, 28, 30, 60, 120, 200, 260, 180, 130, 110,
              100, 105, 110, 130, 200, 260, 240, 180, 130, 100, 70, 50],
})
df.head()
```

## Render

`glyph.render(spec, data=df)` registers the DataFrame as an in-process view,
compiles the spec, and returns a `RenderResult`. The `RenderResult` has a
`_repr_mimebundle_` so the bare `result` cell renders inline in Jupyter.

```python
result = glyph.render(
    {"layers": [{"mark": "bar", "encoding": {"x": "hour", "y": "rides"}}]},
    data=df,
)
result  # renders inline in Jupyter
```

## Audit findings

Glyph runs 8 audit rules (color contrast, data-ink ratio, etc.) on every
render. Empty list = no findings on this chart.

```python
for f in result.audit:
    print(f"[{f.severity}] {f.rule_id}: {f.message}")
print(f"Total findings: {len(result.audit)}")
```
